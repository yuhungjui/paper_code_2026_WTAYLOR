# Visualization: Wasserstein Taylor Diagram 

- Prepare input values.
- Plot basic Wasserstein Taylor Diagram as example.


```python
import numpy as np

np.random.seed(42)

n = 500

# Observed data: exact sample mean and sample std
mu_obs = 1
std_obs = 2.3

z_obs = np.random.normal(size=n)
z_obs = (z_obs - np.mean(z_obs)) / np.std(z_obs, ddof=0)

observed = mu_obs + std_obs * z_obs

# Predicted data settings
means_pred = [0.5, 1.2, 2.0, -0.5, 1.8]
stds_pred  = [2, 2, 3, 4, 4.5]
corrs_pred = [0.4, 0.95, 0.7, 0.9, 0.5]

predicted_datasets = []

for mu_pred, std_pred, rho in zip(means_pred, stds_pred, corrs_pred):
    noise = np.random.normal(size=n)

    # Make noise orthogonal to z_obs
    noise = noise - np.mean(noise)
    noise = noise - np.dot(noise, z_obs) / np.dot(z_obs, z_obs) * z_obs
    noise = noise / np.std(noise, ddof=0)

    # Construct variable with exact sample correlation rho
    z_pred = rho * z_obs + np.sqrt(1 - rho**2) * noise

    pred = mu_pred + std_pred * z_pred
    predicted_datasets.append(pred)

# Check
print("Observed:")
print(np.mean(observed), np.std(observed, ddof=0))

for i, pred in enumerate(predicted_datasets):
    print(f"Dataset {i+1}:")
    print("mean =", np.mean(pred))
    print("std  =", np.std(pred, ddof=0))
    print("corr =", np.corrcoef(observed, pred)[0, 1])
```

```python
import pandas as pd
df = pd.read_csv("M2_wind_spead_2010_2020.csv")
df.head()
Wind_Speed_Obs = df["Avg Avg Wind Speed @ 80m [m/s]"]
Wind_Speed_Obs.shape
# Plot histogram
freq, bins, _ = plt.hist(Wind_Speed_Obs, bins=80)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Wind Speed")
plt.show()
```

```python
import numpy as np
from scipy import stats
from scipy.special import betainc
from scipy.optimize import brentq

x = np.asarray(Wind_Speed_Obs)
n = len(x)
u = (np.arange(1, n + 1) - 0.5) / n

x_sorted = np.sort(x)
x_centered = x_sorted - np.mean(x)


def centered_w2_distance(x_centered, q):
    q_centered = q - np.mean(q)
    return np.sqrt(np.mean((x_centered - q_centered) ** 2))


def quantile_corr(x_centered, q):
    q_centered = q - np.mean(q)
    return np.mean(x_centered * q_centered) / (
        np.std(x_centered, ddof=0) * np.std(q_centered, ddof=0)
    )


# --------------------------------------------------
# Basic CDFs
# --------------------------------------------------
def lindley_cdf(x, lam):
    return 1 - (1 + lam * x / (1 + lam)) * np.exp(-lam * x)


def generalized_lindley_cdf(x, alpha, lam):
    return lindley_cdf(x, lam) ** alpha


def beta_generated_cdf(G, a, b):
    return betainc(a, b, G)


# --------------------------------------------------
# Numerical inverse CDF
# --------------------------------------------------
def numerical_ppf(cdf, u, upper_start=0.1):
    q = []

    for ui in u:
        upper = upper_start
        while cdf(upper) < ui:
            upper *= 2

        root = brentq(lambda z: cdf(z) - ui, 0, upper)
        q.append(root)

    return np.array(q)


# --------------------------------------------------
# 80 m parameters from your table
# --------------------------------------------------
params = {
    "BGL":  {"alpha": 25.281, "lam": 0.669, "a": 0.093, "b": 0.456},
    "BE":   {"lam": 0.882, "a": 3.279, "b": 0.353},
    "BW":   {"alpha": 0.463, "lam": 0.175, "a": 9.712, "b": 7.680},
    "BL":   {"lam": 1.415, "a": 2.894, "b": 0.241},
    "GEV":  {"lam": 1.941, "a": 2.945, "b": 0.245},
    "LogN": {"alpha": 1.285, "lam": 0.730},
    "GAM":  {"alpha": 2.158, "lam": 0.465},
    "GL":   {"alpha": 1.625, "lam": 0.472},
    "W":    {"alpha": 1.463, "lam": 0.194},
    "L":    {"lam": 0.373},
}


# --------------------------------------------------
# Quantile functions
# --------------------------------------------------
quantiles = {}

# BGL
p = params["BGL"]
quantiles["BGL"] = numerical_ppf(
    lambda z: beta_generated_cdf(
        generalized_lindley_cdf(z, p["alpha"], p["lam"]),
        p["a"],
        p["b"]
    ),
    u
)

# BE
p = params["BE"]
quantiles["BE"] = numerical_ppf(
    lambda z: beta_generated_cdf(
        1 - np.exp(-p["lam"] * z),
        p["a"],
        p["b"]
    ),
    u
)

# BW
p = params["BW"]
quantiles["BW"] = numerical_ppf(
    lambda z: beta_generated_cdf(
        1 - np.exp(-(p["lam"] * z) ** p["alpha"]),
        p["a"],
        p["b"]
    ),
    u
)

# BL
p = params["BL"]
quantiles["BL"] = numerical_ppf(
    lambda z: beta_generated_cdf(
        lindley_cdf(z, p["lam"]),
        p["a"],
        p["b"]
    ),
    u
)

# GL
p = params["GL"]
quantiles["GL"] = numerical_ppf(
    lambda z: generalized_lindley_cdf(z, p["alpha"], p["lam"]),
    u
)

# L
p = params["L"]
quantiles["L"] = numerical_ppf(
    lambda z: lindley_cdf(z, p["lam"]),
    u
)

# Log-normal
p = params["LogN"]
quantiles["LogN"] = stats.lognorm.ppf(
    u,
    s=p["lam"],
    scale=np.exp(p["alpha"])
)

# Gamma
p = params["GAM"]
quantiles["GAM"] = stats.gamma.ppf(
    u,
    a=p["alpha"],
    scale=1 / p["lam"]
)

# Weibull
p = params["W"]
quantiles["W"] = stats.weibull_min.ppf(
    u,
    c=p["alpha"],
    scale=1 / p["lam"]
)

# GEV
p = params["GEV"]
quantiles["GEV"] = stats.genextreme.ppf(
    u,
    c=-p["b"],
    loc=p["a"],
    scale=p["lam"]
)


# --------------------------------------------------
# Compute centered W2 and quantile correlation
# --------------------------------------------------
for name, q in quantiles.items():
    w2c = centered_w2_distance(x_centered, q)
    qcorr = quantile_corr(x_centered, q)
    print(f"{name:5s}  centered W2 = {w2c:.4f}   quantile corr = {qcorr:.4f}")
```

```python
Centered_W_2 = []
Quantile_Corr = []
Model = []
prediction_std =[]
for name, q in quantiles.items():
    w2c = centered_w2_distance(x_centered, q)
    qcorr = quantile_corr(x_centered, q)
    fitted_std = np.std(q, ddof=0)
    Model.append(name)
    Centered_W_2.append(w2c)
    Quantile_Corr.append(qcorr)
    prediction_std.append(fitted_std)
print(Model)
print(Centered_W_2)
print(Quantile_Corr)
print(prediction_std)
```

```python
# --------------------------------------------------
# Verify the law of cosines
#
# W_{2,c}^2
# =
# sigma_x^2 + sigma_q^2
# - 2 sigma_x sigma_q rho_q
# --------------------------------------------------

sigma_x = np.std(Wind_Speed_Obs, ddof=0)

for i in range(len(Model)):

    sigma_q = prediction_std[i]
    rho_q = Quantile_Corr[i]
    w2c = Centered_W_2[i]

    lhs = w2c**2

    rhs = (
        sigma_x**2
        + sigma_q**2
        - 2 * sigma_x * sigma_q * rho_q
    )

    print(f"Model: {Model[i]}")
    print(f"LHS = W2_c^2      = {lhs:.10f}")
    print(f"RHS law cosine    = {rhs:.10f}")
    print(f"Difference        = {abs(lhs-rhs):.2e}")
    print("-" * 60)
```
